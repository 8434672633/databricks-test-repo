```python
path = '/Volumes/dev/bronze/bronze-volume/day1_sql_practice.sql'

with open(path, 'r') as f:
    lines = f.readlines()

# Filter out Databricks notebook markup (-- MAGIC lines) and extract SQL
sql_lines = []
for line in lines:
    # Skip notebook source header
    if line.strip().startswith('-- Databricks notebook source'):
        continue
    # Skip -- MAGIC lines (markdown and comments)
    if '-- MAGIC' in line:
        continue
    # Keep regular SQL lines
    sql_lines.append(line)

# Join and split by semicolons to get individual statements
script = ''.join(sql_lines)
statements = [stmt.strip() for stmt in script.split(';') if stmt.strip()]

# Execute each SQL statement
for i, stmt in enumerate(statements):
    if stmt:
        print(f"Executing statement {i+1}/{len(statements)}...")
        try:
            spark.sql(stmt)
            print(f"✓ Statement {i+1} completed successfully")
        except Exception as e:
            print(f"✗ Error in statement {i+1}: {e}")
            break
```

```python
from pyspark.sql.functions import col, when
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("parent-master-mapping").getOrCreate()

transactions_df = spark.table("dev.bronze.day1_transaction_raw")
customers_df = spark.table("dev.bronze.day1_customers")
merchants_df = spark.table("dev.bronze.day1_merchants")


customer_lookup_df = customers_df.select(col("customer_id").alias("master_customer_id"))

merchant_lookup_df = merchants_df.select(col("merchant_id").alias("master_merchant_id"))


validated_df = ( transactions_df.alias("t").join(customer_lookup_df.alias("c"),col("t.customer_id") == col("c.master_customer_id"),"left").join(
    merchant_lookup_df.alias("m"),col("t.merchant_id") == col("m.master_merchant_id"), "left").withColumn("customer_issue",when(col("t.customer_id").isNull(),"Missing")
    .when(
        col("c.master_customer_id").isNull(),"UNKNOWN").otherwise("OK")).withColumn("merchant_issue",when(col("t.merchant_id").isNull(),"Missing").when(col("m.master_merchant_id").isNull(),"UNKNOWN").otherwise("OK"))
).select("*").orderBy(col("transaction_id"))
            
                
validated_df.show(5)
```

```python
from pyspark.sql.functions import sum as spark_sum, count

cumm_sum_transaction_df = transactions_df.groupBy(col("transaction_id")).agg(spark_sum(col("amount")).alias("cumm_sum"), count("*").alias("counts")).orderBy(col("transaction_id"))

cumm_sum_transaction_df.show()                                                                
```

```python

from pyspark.sql.functions import dense_rank, sum as spark_sum
from pyspark.sql.window import Window

joined_df = transactions_df.alias("t").join(merchants_df.alias("m"),col("t.merchant_id") == col("m.merchant_id"), "inner")

ranked_df = joined_df.groupby(col("transaction_id")).agg(sum("amount").alias("total_amount"))

w = window.orderby(col("total_amount").desc())
ranked_df_with_rank = ranked_df.withColumn("rank", dense_rank().over(w))
ranked_df_with_rank.show(5)
```
